## Problem Three: Move All Zeros to End
**Description**  
Given a list of integers, move all zeros to the end while maintaining the relative order of the non-zero elements. 

**Examples**
* Example 1:
    ```python
    Input: nums = [0, 1, 0, 3, 12]
    Output: [1, 3, 12, 0, 0]
    ```
* Example 2:
    ```python
    Input: nums = [4, 0, 5, 0, 0, 6]
    Output: [4, 5, 6, 0, 0, 0]
    ```

**Problem 3 Starter Code**

````python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    # TODO
    result = []
    zero_count = 0

    # Go through each number in the list
    for num in nums:
        if num == 0:
            zero_count += 1
        else:
            result.append(num)

    # Add all zeros at the end
    for i in range(zero_count):
        result.append(0)

    return result


# Example 1
nums = [0, 1, 0, 3, 12]
print(move_zeros_to_end(nums))

# Example 2
nums = [4, 0, 5, 0, 0, 6]
print(move_zeros_to_end(nums))
````

3. Based on the provided starter code, code the solution to your assigned problem here in Python. Note: each problem can be solved more simply if you use an abstract data type that is suitable for that problem. Try to find the solution with best time and space complexity.

````python
from typing import List

def move_zeros_to_end(nums: List[int], index=0) -> List[int]:
    ## Base case definition: If the index reaches the length of the list, return an empty list.
    if index == len(nums): 
        return []

   ## Recursive call: Call the function recursively with the next index and store the result in a variable.
    result = move_zeros_to_end(nums, index + 1) 

    if nums[index] != 0:
        result.insert(0, nums[index])

    else:
        result.append(0)

    return result


## Example 1
nums = [0, 1, 0, 3, 12]
print(move_zeros_to_end(nums))

# Example 2
nums = [4, 0, 5, 0, 0, 6]
print(move_zeros_to_end(nums))
````
