## Problem Three: Missing Number in Range
**Description**  
You are given a list containing `n` non-negative integers in the range `[1, n]`. Return a list of numbers that are missing from the range `[1, n]` in the given list. If there is no missing number, return -1. 

**Note**: integers in the list may not be unique.

**Examples**

* Example 1:

    Input: `lst = [1, 1, 3]`

    Output: [2]

* Example 2:

    Input: `lst = [5,3,2,4,1]`

    Output: -1

* Example 3:

    Input: `lst = [6, 8, 2, 3, 5, 7, 1, 1, 2, 10]`

    Output: [4, 9]

**Problem 3 Starter Code**

```python
from typing import List
def missing_num(nums: List) -> int:
  # Code your solution here #
  if len(nums) == 0 or len(nums) == 1:
    return -1
  else:
    n = len(nums)
    missing_number = []
    for i in range(1, n + 1):
      if i not in nums:
        missing_number.append(i)
 
    if len(missing_number) == 0:
      return -1
    else:
      return missing_number
    
# Example 1
nums_example1 = [1, 1, 3]
print(missing_num(nums_example1))  #  Output: [2]
  
```

```python
# Example 2
nums_example2 = [5,3,2,4,1]
print(missing_num(nums_example2))  #  Output: -1
```
